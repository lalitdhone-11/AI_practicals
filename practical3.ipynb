```python
a=[5,2,7,1,3,4]
def selection_Sort(a):
    for i in range(len(a)-1):
        min_index=i
        for j in range(i,len(a)):
            if(a[min_index]>a[j]):
                min_index=j
        a[min_index],a[i]=a[i],a[min_index]
    
print("unsorted list",a)
selection_Sort(a)
print("sorted list",a)
```

```python
import heapq
def prims(g,s):
    vis=set()
    min_heap=[(0,s)]
    total_cost=0
    mst_edge=[]
    while min_heap:
        weight,node=heapq.heappop(min_heap)
        if node in vis:
            continue
        vis.add(node)
        total_cost += weight
        for neigh,e_w in g[node]:
            if neigh not in vis:
                heapq.heappush(min_heap,(e_w,neigh))
                mst_edge.append((node,neigh,e_w))
    return total_cost ,mst_edge

g = {
    'A': [('B',2), ('C',3)],
    'B': [('A',2), ('C',1), ('D',4)],
    'C': [('A',3), ('B',1), ('D',5)],
    'D': [('B',4), ('C',5)]
}
cost,edge=prims(g,'A')
print("mst cost:",cost)
print("mst egdes:",edge)
```
