```python
from  collections import deque

#DFS
g = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D', 'E'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C'],
    'E': []
}
def dfsrev(g,s):
    vis.add(s)
    print(s)
    for c in g[s]:
        if c not in vis:
            dfsrev(g,c)
#vis=set()
#dfsrev(g,'A')

#BFS
def bfsrev(q,vis,g):
    if not q:
        return
    cur =q.popleft()
    print(cur,end=" ")
    for c in g[cur]:
            if c not in vis:
                q.append(c)
                vis.add(c)
    bfsrev(q,vis,g)
q=deque(['A'])
vis=set(['A'])
bfsrev(q,vis,g)
    
```
